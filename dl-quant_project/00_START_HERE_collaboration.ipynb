## Step 2 · Check your setup

Run the cell below. It mounts your Drive, finds the shared folder (wherever your shortcut is), and lists the data files.

````python
from google.colab import drive
drive.mount('/content/drive')

# ------------------------------------------------------------------
# Project paths: shared Google Drive folder "dl-quant_project"
# Works for the owner and for teammates who added a shortcut to the
# shared folder in their My Drive (see 00_START_HERE_collaboration).
# ------------------------------------------------------------------
import os, glob

def find_project_dir(name="dl-quant_project"):
    candidates = [
        f"/content/drive/MyDrive/Colab Notebooks/{name}",
        f"/content/drive/MyDrive/{name}",
        *glob.glob(f"/content/drive/MyDrive/*/{name}"),
        *glob.glob(f"/content/drive/Shareddrives/*/{name}"),
    ]
    for p in candidates:
        if os.path.isdir(p):
            return p
    raise FileNotFoundError(
        f"Could not find the '{name}' folder in your Google Drive.\n"
        "Fix: Google Drive > Shared with me > right-click 'dl-quant_project' > "
        "Organize > Add shortcut > My Drive. Then re-run this cell.")

PROJECT_DIR  = find_project_dir()
DATA_DIR     = os.path.join(PROJECT_DIR, "DataSource")       # all data files: read AND save here
NOTEBOOK_DIR = os.path.join(PROJECT_DIR, "PythonNoteBooks")  # notebooks
os.makedirs(DATA_DIR, exist_ok=True)
drive_folder = DATA_DIR   # older cells use this name
print("Project folder:", PROJECT_DIR)
print("Data folder:   ", DATA_DIR)

print("\nFiles in DataSource:")
for f in sorted(os.listdir(DATA_DIR)):
    size_mb = os.path.getsize(os.path.join(DATA_DIR, f)) / 1e6
    print(f"  {f:40s} {size_mb:8.1f} MB")
````

## Rules for working together

**Notebooks**
- **One owner per notebook** (see the table below). Others can read and run it, but edit only after telling the owner.
- To try changes, use **File → Save a copy in Drive**, then move the copy into `PythonNoteBooks/` with your name in it, e.g. `02_keyword_baseline_sentiment_ALEX.ipynb`.
- Colab is **not** like Google Docs: two people editing the same notebook at the same time can overwrite each other. If something is lost, use **File → Revision history**.
- Each person who clicks *Run* gets their **own** runtime: variables and files in `/content/` are not shared. Only files saved to `DATA_DIR` are shared.
- To let others see your results, turn off **Edit → Notebook settings → "Omit code cell output when saving"** and don't use *private outputs*. Also write a short **Key takeaways** cell at the end.

**Data files**
- Save every data file to `DATA_DIR` only (no local copies passed around).
- File names: `<what>_<source>_<version>.csv`, e.g. `daily_signals_finbert_v1.csv`. **Never reuse an existing file name** unless you own that file. A save silently replaces it for everyone.
- Large raw downloads (e.g. the 23 GB FNSPID file) stay in Colab's temporary storage. Do **not** copy them to the shared folder (Drive space is limited).
- If you add or change a data file, add a line to the catalog below.
- New file not showing up? The person who saved it should run `drive.flush_and_unmount()`. You can re-mount with `drive.mount('/content/drive', force_remount=True)`.

## Troubleshooting

| Problem | Fix |
|---|---|
| `FileNotFoundError: Could not find the 'dl-quant_project' folder` | You have no shortcut yet. Do Step 1, then re-run the setup cell. |
| `FileNotFoundError` for a CSV file | The file isn't in `DataSource/` yet. Run the notebook that writes it (see the catalog), or ask its owner. |
| `PermissionError` / "Read-only file system" when saving | You have Viewer access. Ask Bianca for Editor access. |
| Setup cell finds the wrong folder | You have your own folder named `dl-quant_project`. Rename or delete it, keep only the shortcut. |
| A teammate's new file doesn't appear | Wait a minute, or re-mount: `drive.mount('/content/drive', force_remount=True)`. |
| "Notebook has changed" warning | Someone else saved the notebook. Reload the page before editing; don't overwrite their version. |
| Session disconnected, variables gone | Normal in Colab. Re-run from the setup cell. Anything saved to `DATA_DIR` is still there. |
